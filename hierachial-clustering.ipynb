```python
import matplotlib.pyplot as plt

from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.cluster import AgglomerativeClustering

from scipy.cluster.hierarchy import linkage, dendrogram


# -----------------------------
# 1. Documents
# -----------------------------

documents = [
    "Python programming machine learning algorithms",
    "Deep learning neural networks artificial intelligence",
    "Football cricket tennis sports players",
    "Cricket football match tournament players",
    "Python pandas numpy machine learning"
]


# -----------------------------
# 2. TF-IDF
# -----------------------------

vectorizer = TfidfVectorizer()

X = vectorizer.fit_transform(documents)

print("TF-IDF Matrix Shape:", X.shape)


# -----------------------------
# 3. Dendrogram
# -----------------------------

Z = linkage(X.toarray(), method="ward")

plt.figure(figsize=(10, 6))

dendrogram(
    Z,
    labels=[f"D{i+1}" for i in range(len(documents))]
)

plt.title("Document Clustering Dendrogram")
plt.xlabel("Documents")
plt.ylabel("Distance")

plt.show()


# -----------------------------
# 4. Agglomerative Clustering
# -----------------------------

model = AgglomerativeClustering(
    n_clusters=2,
    linkage="ward"
)

labels = model.fit_predict(X.toarray())


# -----------------------------
# 5. Display Results
# -----------------------------

for i, label in enumerate(labels):

    print(
        f"Document {i+1} → Cluster {label}"
    )
```
