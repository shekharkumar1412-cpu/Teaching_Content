```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import tensorflow as tf
import keras
from sklearn.datasets import make_moons
from sklearn.model_selection import train_test_split
from keras.layers import Dense
from keras.models import Sequential
```

### Making the classification dataset using make_moons function

```python
X,y = make_moons(n_samples=250, noise=0.05, random_state=42)
```

```python
# plotting the data 
plt.scatter(X[:,0],X[:,1], c=y, s=100)
plt.show()
```

# Creating the Neural Network using sequential and Dense 

```python
model = Sequential()

model.add(Dense(10,activation='relu',input_dim=2))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(10,activation='relu'))
model.add(Dense(1, activation='sigmoid'))
```

# compiling the model

```python
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

# initial weights for hidden layer 1

```python
model.get_weights()[0]
```

```python
# storing the initial weights in a variable

old_weights = model.get_weights()[0]
```

# Train test split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.20, random_state=42)
```

# training the model and finding the new weights

```python
model.fit(X_train, y_train, epochs = 100)
```

```python
new_weights = model.get_weights()[0]
```

# finding the learning rate and gradient and percent change in the weights 

```python
model.optimizer.get_config()["learning_rate"]
```

```python
gradient = (old_weights - new_weights)/ 0.001
percent_change = abs(100*(old_weights - new_weights)/ old_weights)
```
