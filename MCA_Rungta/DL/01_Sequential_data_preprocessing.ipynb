```python
import numpy as np
```

```python
# input text data

sentences = ["the cat sat on the mat",
             "the dog barked",
             "a cat and a dog played in the garden"]
```

```python
# Tokenize and build a vocabulary (index 0 reserved for padding)
tokens = [s.split() for s in sentences]
vocab = {"<pad>": 0, "<unk>": 1}
for sent in tokens:
    for w in sent:
        vocab.setdefault(w, len(vocab))
```

```python
print("Vocabulary size:", len(vocab))
print('-'*70)
print('Created vocabulory where we assigned a number to each tokens:')
vocab
```

#### Convert each sentence to a sequence of integer ids


```python
ids = [[vocab.get(w, vocab["<unk>"]) for w in sent] for sent in tokens]
ids
 
```

##### Pad to equal length so sequences can be batched, and build a mask


```python
max_len = max(len(s) for s in ids)
padded = np.array([s + [0] * (max_len - len(s)) for s in ids])
mask = (padded != 0).astype(float)          # 1 = real token, 0 = padding
lengths = mask.sum(axis=1).astype(int)
```

```python
print("Padded batch shape (batch, time):", padded.shape)
print(padded)
print("Original text Sequence's lengths:", lengths)
```

####  One-hot representation: (batch, time, vocab)


```python
one_hot = np.eye(len(vocab))[padded]
print("One-hot tensor shape:", one_hot.shape)
 
```
