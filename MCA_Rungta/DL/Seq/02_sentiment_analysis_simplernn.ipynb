# Input data in text format


```python
docs = ['go india',
		'india india',
		'hip hip hurray',
		'jeetega bhai jeetega india jeetega',
		'bharat mata ki jai',
		'kohli kohli',
		'sachin sachin',
		'dhoni dhoni',
		'nehru ji ki jai',
		'inquilab zindabad']
```

# Implementing integer encoding

```python
from tensorflow.keras.preprocessing.text import Tokenizer
```

```python
tokenizer = Tokenizer()
tokenizer.fit_on_texts(docs)
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

```python
sequences = tokenizer.texts_to_sequences(docs)
sequences
```

```python
from keras.utils import pad_sequences
sequences = pad_sequences(sequences,padding='post')
sequences
```

```python
print("Sequences shape:", sequences.shape)
print("Model input shape:", model.input_shape)
print("Sequences dtype:", sequences.dtype)
print("Maximum token:", sequences.max())
print("Minimum token:", sequences.min())
```

### Implementing embedding to create a dense vector from integer encoded inputs

```python
from keras.models import Sequential
from keras.layers import Embedding,Input
```

```python
model = Sequential()
model.add(Input(shape=(5,)))
# 17 is total unique words in our vocab
model.add(Embedding(input_dim=18,output_dim=2))

model.summary()
```

```python
model.compile('adam','accuracy')
```

```python
pred = model.predict(sequences)
print(pred)
```

#### now above we had seen how we create embeddings now using imbd data and building model on it

```python
from keras.datasets import imdb
from tensorflow.keras.preprocessing.text import Tokenizer
from keras.utils import pad_sequences
from keras import Sequential
from keras.layers import Dense,SimpleRNN,Embedding,Flatten
```

```python
(X_train,y_train),(X_test,y_test) = imdb.load_data()
```

```python
X_train = pad_sequences(X_train,padding='post',maxlen=50)
X_test = pad_sequences(X_test,padding='post',maxlen=50)
```

```python
print(X_train.shape)
print(X_test.shape)
```

```python
model = Sequential()

model.add(Input(shape=(50,)))  # Each sequence has 50 tokens
model.add(Embedding(input_dim=88586, output_dim=2))
model.add(SimpleRNN(32, return_sequences=False))
model.add(Dense(1, activation='sigmoid'))

model.summary()
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])

```

```python
history = model.fit(X_train, y_train,epochs=5,validation_data=(X_test,y_test))
```
