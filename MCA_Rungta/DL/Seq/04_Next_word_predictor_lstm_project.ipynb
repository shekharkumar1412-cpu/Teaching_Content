```python
import tensorflow
from tensorflow.keras.preprocessing.text import Tokenizer
```

# implementing tokenizer which will assign a unique value to each unique words tokens

```python
tokenizer = Tokenizer()
```

```python
tokenizer.fit_on_texts([text_data])
```

```python
tokenizer.word_index
```

```python
len(tokenizer.word_index)
```

**Above using tokenizer we assigned a unique number to each word**

# Now we have to create input and output data sentence wise


```python
# Fteching out all sentence from our text data

for sentence in text_data.split('\n'):
  print(sentence)
```

```python
# generating input sequence for each sentence (to create labelled data)

input_sequences = []
for sentence in text_data.split('\n'):
  tokenized_sentence = tokenizer.texts_to_sequences([sentence])[0]

  for i in range(1,len(tokenized_sentence)):
    input_sequences.append(tokenized_sentence[:i+1])
```

```python
input_sequences
```

##### Fetching out the length of longest input sentence n order to mae all input sentence of same dienson by implmenting zero padding

```python
max_len = max([len(x) for x in input_sequences])
max_len
```

# Now implementing zero padding

```python
from tensorflow.keras.preprocessing.sequence import pad_sequences
padded_input_sequences = pad_sequences(input_sequences, maxlen = max_len, padding='pre')
```

```python
padded_input_sequences
```

```python
X = padded_input_sequences[:,:-1]
```

```python
y = padded_input_sequences[:,-1]
```

```python
X.shape
```

```python
y.shape
```

# Now we have labelled data in form of numbers



#### Implementing one hot encoding on target variable to convert the problem into multi class classification

```python
from tensorflow.keras.utils import to_categorical
y = to_categorical(y,num_classes=102)
```

```python
y.shape
```

### Building a LSTM architecture

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Embedding, LSTM, Dense, Input

model = Sequential([
    Input(shape=(10,), dtype="int32"),

    # adding a embedding layer to make the sparse vector into a dense vector
    # in embedding we create a 100 dimensional vector for each words
    Embedding(input_dim=102, output_dim=100),

    # each neural layer will have 150 nodes in the lstm architecure
    LSTM(150),

    # output layer
    Dense(102, activation="softmax")
])

model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()
```

```python
model.fit(X,y,epochs=100)
```

# Prediction on new data

```python
import time
import numpy as np


text = "natural"

for i in range(6):
  # tokenize
  token_text = tokenizer.texts_to_sequences([text])[0]
  # padding
  padded_token_text = pad_sequences([token_text], maxlen=10, padding='pre')
  # predict
  pos = np.argmax(model.predict(padded_token_text))

  for word,index in tokenizer.word_index.items():
    if index == pos:
      text = text + " " + word
      print(text)
      time.sleep(2)
```
