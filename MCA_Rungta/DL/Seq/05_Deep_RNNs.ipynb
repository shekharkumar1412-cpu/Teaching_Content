```python
import tensorflow as tf
from tensorflow.keras.datasets import imdb
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Input , Embedding, SimpleRNN, Dense,LSTM,GRU
```

```python
# Load the IMDb dataset with forcing Keras to keep only the 10,000 most frequent words in its word-index vocabulary
(x_train, y_train), (x_test, y_test) = imdb.load_data(num_words=10000)

# Pad sequences to have the same length short reviews will be padded with zeros and long reviews are truncated to 100 tokens.
x_train = pad_sequences(x_train, maxlen=100)
x_test = pad_sequences(x_test, maxlen=100)
```

```python
# Define the RNN model
model = Sequential([
    Input(shape=(100,), dtype="int32"),

    # Convert each of 10k words to a dense numerical vector of 32 dimension
    Embedding(input_dim = 10000, output_dim = 32),

    SimpleRNN(5, return_sequences=True),   # RNN layer with 5 units
    SimpleRNN(5),                          # Another RNN layer with 5 units

    # Output layer for binary classification
    Dense(1, activation='sigmoid')         
])

model.summary()
```

```python
# Define the LSTM model
model = Sequential([
    Input(shape=(100,), dtype="int32"),  
    Embedding(input_dim = 10000, output_dim = 32),
    
    LSTM(5, return_sequences=True),
    LSTM(5),
    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
# Define the GRU model
model = Sequential([
    Input(shape=(100,), dtype="int32"),
    Embedding(input_dim = 10000, output_dim = 32),
    GRU(5, return_sequences=True),
    GRU(5),
    Dense(1, activation='sigmoid')
])

model.summary()
```

```python
# Compile the model
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
# Train the model
history = model.fit(x_train, y_train, epochs=5, batch_size=32, validation_split=0.2)
```
