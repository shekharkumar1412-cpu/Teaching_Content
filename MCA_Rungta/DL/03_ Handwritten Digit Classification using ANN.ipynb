# MNIST Dataset

* It is a collection of over 70k images of handwritten numbers

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense,Flatten
```

```python
# Loading the mnist inbuilt datset from keras 
(X_train,y_train),(X_test,y_test) = keras.datasets.mnist.load_data()
```

```python
# X values are in range of 0 to 255 so scaling them to a scale of 0 to 1 
# pixel values ranging from 0 to 255 because standard digital images allocate 1 byte (8 bits) of memory
# to store the brightness level of each pixel

X_train = X_train/255
X_test = X_test/255
```

```python
X_train[0]
```

```python
# Building the model 

model = Sequential()

# input layer 784 input features 
model.add(Flatten(input_shape=(28,28)))

# first hidden layer having 128 neurons
model.add(Dense(128,activation='relu'))

# second hidden layers having 32 neurons
model.add(Dense(32,activation='relu'))

# output layer having 10 neurons as 10 categories are there for classification
model.add(Dense(10,activation='softmax'))
```

```python
model.summary()
```

```python
model.compile(loss='sparse_categorical_crossentropy',optimizer='Adam',metrics=['accuracy'])




# model.compile(loss ='mean_squared_error' , optimizer ='Adam') for regr
```

* In Keras/TensorFlow, if you don't specify batch_size, it does NOT mean Batch Gradient Descent.

* For model.fit(), the default is: batch_size=32

```python
history = model.fit(X_train,y_train,epochs=25,validation_split=0.2)
```

```python
y_prob = model.predict(X_test)
y_pred = y_prob.argmax(axis=1)
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
