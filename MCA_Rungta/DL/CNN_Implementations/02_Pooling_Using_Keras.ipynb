```python
import tensorflow
from tensorflow import keras
from keras.layers import Dense,Conv2D,Flatten,MaxPooling2D
from keras import Sequential
from keras.datasets import mnist
```

```python
(X_train, y_train), (X_test, y_test) = mnist.load_data()
```

```python
model = Sequential()

# Adding first convolution layer having 32 filters/kernel each of size 3 by 3 and there is no padding here
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu', input_shape=(28,28,1)))

# now after first convolution layer adding a pooling layer
model.add(MaxPooling2D(pool_size=(2, 2), strides=2, padding='valid'))

# Adding the seconnd convolution layer in our architecture
model.add(Conv2D(32,kernel_size=(3,3),padding='valid', activation='relu'))

# now again adding a pooling layer after second convolution layer
model.add(MaxPooling2D(pool_size=(2, 2), strides=2, padding='valid'))

# now we are just converting the 2 dimensional feature map into a one dimensional
#so that it can be passed to fully connected layer
model.add(Flatten())

# Building the fully connected layer of CNN
model.add(Dense(128,activation='relu'))
model.add(Dense(10,activation='softmax'))
```
