```python
from tensorflow.keras.preprocessing import image
from tensorflow.keras.preprocessing.image import ImageDataGenerator
```

**Loading the image using image module rather than opencv**

```python
img = image.load_img("cat.100.jpg",target_size = (250,250))
img
```

**Here the datatype of loaded image is not a numpy array unlike opencv**

```python
type(img)
```

**converting loaded image into numpy array numbers(pixels)**

```python
# converting image to array 

img = image.img_to_array(img)
img.shape
```

```python
type(img)
```

```python
input_batch = img.reshape(1,250,250,3)
```

# Implementing Data Augmentation using ImageDataGenerator

**refer to this blog**

https://blog.keras.io/building-powerful-image-classification-models-using-very-little-data.html 

```python
datagen = ImageDataGenerator(
        rotation_range=40,   # refer to angle of rotation
        shear_range=0.2,
        zoom_range=0.2,
        horizontal_flip=True,
        width_shift_range=0.2,
        height_shift_range=0.2,
        fill_mode='constant')
```

# Creating new image and storing it 

```python
i=0
for output in datagen.flow(input_batch,
                               batch_size = 1,
                               save_to_dir = 'Generated_Img'):


    i = i+1
    if i == 20:
        break
```
