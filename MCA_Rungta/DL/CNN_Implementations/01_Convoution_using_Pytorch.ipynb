```python
import torch
import torch.nn as nn
                                                                                                                                                                                                                                                                                                                                                                  
# Batch size , channel , width height 
x = torch.randn(1, 3, 32, 32)
```

```python
conv = nn.Conv2d(
    in_channels=3,
    out_channels=16,
    kernel_size=3,
    stride=1,
    padding=1
)
```

```python
y = conv(x)

print("Input :", x.shape)
print("Output:", y.shape)
```
