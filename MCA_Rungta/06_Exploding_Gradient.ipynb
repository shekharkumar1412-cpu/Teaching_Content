```python
import numpy as np
```

```python
np.random.seed(0) 

# setting the number of layers = 15
num_layers = 15 

# initial gradient 
grad = np.ones((1, 10))

grad
```

```python
gradient_norms = [] 

# Deliberately large weight scale to trigger explosion 
weights = [np.random.randn(10, 10) * 2.5 for _ in range(num_layers)] 

weights
```

```python
for W in reversed(weights): 
    grad = grad @ W.T   # no squashing activation derivative here 
    gradient_norms.append(np.linalg.norm(grad)) 
print("Gradient norm at each layer:") 
for i, g in enumerate(gradient_norms): 
    print(f"Layer {i+1}: {g:.4f}") 
```
