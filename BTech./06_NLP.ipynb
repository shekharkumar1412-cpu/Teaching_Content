```python
import nltk 
from nltk.tokenize import word_tokenize, sent_tokenize 
```

```python
text = """Natural Language Processing is amazing! It
helps computers understand human language. Dr. Smith uses NLP for research."""
```

```python
# Sentence Tokenization 
sentences = sent_tokenize(text) 
print("Sentence Tokens:") 
for i, s in enumerate(sentences, 1): 
    print(i, '-', s) 
```

```python
 # Word Tokenization 
words = word_tokenize(text) 
print("\nWord Tokens:") 
print(words) 
print("\nTotal Words:", len(words)) 
```

# Stop Word Removal 

```python
from nltk.corpus import stopwords 
from nltk.tokenize import word_tokenize 
text = "This is a simple example to show how stopword removal works in NLP" 
tokens = word_tokenize(text) 
stop_words = set(stopwords.words('english')) 
filtered_tokens = [w for w in tokens if w.lower() not in stop_words] 
print("Original Tokens:", tokens) 
print("Number of Stopwords in English:", len(stop_words)) 
print("Filtered Tokens:", filtered_tokens)
```

# Stemming

```python
from nltk.stem import PorterStemmer, SnowballStemmer, LancasterStemmer 
from nltk.stem import WordNetLemmatizer 
    
words = ['maximum', 'frienships', 'organization ', 'generously', 'conditional', 
         'relational', 'fairly', 'sportingly','plays'] 
  
porter = PorterStemmer() 
snowball = SnowballStemmer('english') 
lancaster = LancasterStemmer() 
lemmatizer = WordNetLemmatizer() 

print(f"{'Word':<14}{'Porter':<14}{'Snowball':<14}{'Lancaster':<14}{"Lematizing":<14}") 
for w in words: 
    print(f"{w:<14}{porter.stem(w):<14}{snowball.stem(w):<14}{lancaster.stem(w):<14}{lemmatizer.lemmatize(w):<14}")
```

# BoW(Bag of Words)

```python
from sklearn.feature_extraction.text import CountVectorizer 
import pandas as pd 
corpus = [ 
'the cat sat on the mat', 
'the dog sat on the log', 
'cats and dogs are great pets' 
] 
  
vectorizer = CountVectorizer() 
bow_matrix = vectorizer.fit_transform(corpus)
```

```python
print("Vocabulary:", vectorizer.get_feature_names_out()) 
print("\nBag-of-Words Matrix:") 
df = pd.DataFrame(bow_matrix.toarray(), columns=vectorizer.get_feature_names_out()) 
print(df)   
```

# TF-IDF

```python
from sklearn.feature_extraction.text import TfidfVectorizer 
import pandas as pd 
corpus = [ 
'the cat sat on the mat', 
'the dog sat on the log', 
'cats and dogs are great pets' 
] 
tfidf_vectorizer = TfidfVectorizer() 
tfidf_matrix = tfidf_vectorizer.fit_transform(corpus) 
```

```python
df = pd.DataFrame( 
tfidf_matrix.toarray(), 
columns=tfidf_vectorizer.get_feature_names_out() 
) 
print("TF-IDF Matrix (rounded to 3 decimals):") 
print(df.round(3)) 
```
